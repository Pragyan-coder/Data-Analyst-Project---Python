## Extract information from the given web site
You will extract the data from the below web site: <br> 


```python
#this url contains the data you need to scrape
url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/IBM-DA0321EN-SkillsNetwork/labs/datasets/Programming_Languages.html"
```

The data you need to scrape is the **name of the programming language** and **average annual salary**.<br> It is a good idea to open the url in your web broswer and study the contents of the web page before you start to scrape.


Import the required libraries


```python
# Your code here
import requests
from bs4 import BeautifulSoup
import pandas as pd
```

Download the webpage at the url


```python
data = requests.get(url).text
```

```python
# Send GET request
response = requests.get(url)
soup = BeautifulSoup(response.text, "html.parser")
```

```python
# Find the table
table = soup.find("table")
rows = table.find_all("tr")

# Prepare lists
languages = []
salaries = []
```

```python
# Loop through rows (skip header)
for row in rows[1:]:
    cols = row.find_all("td")
    if len(cols) >= 3:  # language and salary are present
        language = cols[1].text.strip()
        salary = cols[3].text.strip()
        languages.append(language)
        salaries.append(salary)
```

```python
table = soup.find('table')
popular_languages = []

for row in table.find_all('tr'): 
    
    cols = row.find_all('td') 
    language = cols[1].getText() 
    avg_salary = cols[3].getText() 
    popular_languages.append(str(language) + ' , ' + str(avg_salary).replace('$','').replace(',',''))

popular_languages
```

Scrape the `Language name` and `annual average salary`.


```python
#your code goes here
# Create DataFrame
df = pd.DataFrame({"Language": languages, "Average Annual Salary": salaries})

# Display result
print(df)
```

Save the scrapped data into a file named *popular-languages.csv*


```python
# your code goes here
# Save to CSV file
df.to_csv("popular-languages.csv", index=False)

print("Data successfully saved to 'popular-languages.csv'")
```
