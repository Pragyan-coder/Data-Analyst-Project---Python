#### Step 7: Analyzing Programming Language Trends by Region


Analyze the popularity of programming languages by region. Use the `LanguageHaveWorkedWith` column to investigate which languages are most used in different regions.

- Filter data by country or region.

- Visualize the top programming languages by region with a bar plot or heatmap.


```python
# Example: Filter for United States
df_us = df[df['Country'] == 'United States of America']

df_uk = df[df['Country'] == 'United Kingdom of Great Britain and Northern Ireland']

df_norway = df[df['Country'] == 'Norway']
```

```python
# Function to split and count languages
def count_languages(df_region):
    lang_series = df_region['LanguageHaveWorkedWith'].dropna().str.split(';')
    lang_flat = [lang.strip() for sublist in lang_series for lang in sublist]
    return pd.Series(lang_flat).value_counts()

# Example counts for US
us_lang_counts = count_languages(df_us)
print(us_lang_counts.head(10))
```

```python
uk_lang_counts = count_languages(df_uk)
nrwy_lang_counts = count_languages(df_norway)

# Combine into a DataFrame for comparison
lang_trends = pd.DataFrame({
    'United States': us_lang_counts,
    'United Kingdom': uk_lang_counts,
    'Norway': nrwy_lang_counts
}).fillna(0)

print(lang_trends.head(10))
```

```python
# Plot top 5 languages in each region
lang_trends.head(5).plot(kind='bar', figsize=(10,6))
plt.title("Top Programming Languages by Region")
plt.ylabel("Number of Respondents")
plt.xlabel("Programming Language")
```

```python
# Function to split and count languages by region
def count_languages_by_region(df, region):
    df_region = df[df['Country'] == region]
    lang_series = df_region['LanguageHaveWorkedWith'].dropna().str.split(';')
    lang_flat = [lang.strip() for sublist in lang_series for lang in sublist]
    return pd.Series(lang_flat).value_counts()

# Example regions
us_lang_counts = count_languages_by_region(df, 'United States of America')
uk_lang_counts = count_languages_by_region(df, 'United Kingdom of Great Britain and Northern Ireland')
nrwy_lang_counts = count_languages_by_region(df, 'Norway')

# Combine into one DataFrame
lang_trends = pd.DataFrame({
    'United States': us_lang_counts,
    'United Kingdom': uk_lang_counts,
    'Norway': nrwy_lang_counts
}).fillna(0)
```

```python
plt.figure(figsize=(10,6))
sns.barplot(
    x=us_lang_counts.head(10).values,
    y=us_lang_counts.head(10).index,
    palette="viridis"
)

plt.title("Top 10 Programming Languages in United States")
plt.xlabel("Number of Respondents")
plt.ylabel("Programming Language")
```

```python
# Normalize to percentages for fair comparison
lang_trends_percent = lang_trends.div(lang_trends.sum(axis=0), axis=1) * 100

plt.figure(figsize=(12,8))
sns.heatmap(lang_trends_percent.head(10), annot=True, fmt=".1f", cmap="YlGnBu")

plt.title("Top Programming Languages by Region (Percentage of Respondents)")
plt.xlabel("Region")
plt.ylabel("Programming Language")
```
