### Step 1: Import Libraries and Load Data

```python
# Import necessary libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the Stack Overflow survey dataset
data_url = 'https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv'
df = pd.read_csv(data_url)

# Display the first few rows of the dataset
df.head()
```

### Step 3: Handle Missing Data


- Identify missing values in the dataset.

- Impute or remove missing values as necessary to ensure data completeness.



```python
# Count missing values per column
missing_counts = df.isnull().sum().sort_values(ascending=False)

# Display top 15 columns with most missing values
print(missing_counts.head(15))
```

```python
categorical_cols = ['Employment', 'RemoteWork', 'JobSat']
for col in categorical_cols:
    df[col] = df[col].fillna('Unknown')
```

```python
text_cols = ['LanguageHaveWorkedWith', 'DatabaseHaveWorkedWith']
for col in text_cols:
    df[col] = df[col].fillna('None')
```

### Step 5: Visualize Job Satisfaction (Focus on JobSat)


- Create a pie chart or KDE plot to visualize the distribution of `JobSat`.

- Provide an interpretation of the plot, highlighting key trends in job satisfaction.


```python
# Drop missing values
df_job_sat = df.dropna(subset=['JobSat'])

# Count values
job_sat_counts = df_job_sat['JobSat'].value_counts()

# Pie chart
plt.figure(figsize=(8,8))
plt.pie(
    job_sat_counts.values,
    labels=job_sat_counts.index,
    autopct='%1.1f%%',
    startangle=140,
    colors=plt.cm.Paired.colors
)

plt.title("Job Satisfaction Distribution")
```

```python
# Map JobSat categories to numeric scores
job_sat_map = {
    'Very dissatisfied': 1,
    'Dissatisfied': 2,
    'Neutral': 3,
    'Satisfied': 4,
    'Very satisfied': 5
}

# Create the numeric column
df['JobSatScore'] = df['JobSat'].map(job_sat_map)
```

### Step 6: Programming Languages Analysis


- Compare the frequency of programming languages in `LanguageHaveWorkedWith` and `LanguageWantToWorkWith`.
  
- Visualize the overlap or differences using a Venn diagram or a grouped bar chart.


```python
# Function to split and count languages
def count_languages(series):
    lang_series = series.dropna().str.split(';')
    lang_flat = [lang.strip() for sublist in lang_series for lang in sublist]
    return pd.Series(lang_flat).value_counts()

# Count frequencies
lang_have_counts = count_languages(df['LanguageHaveWorkedWith'])
lang_want_counts = count_languages(df['LanguageWantToWorkWith'])
```

```python
lang_compare = pd.DataFrame({
    'HaveWorkedWith': lang_have_counts,
    'WantToWorkWith': lang_want_counts
}).fillna(0).astype(int)

print(lang_compare.head(10))
```

```python
import matplotlib.pyplot as plt

# Select top 10 languages by combined frequency
top_langs = lang_compare.sum(axis=1).sort_values(ascending=False).head(10)
lang_top = lang_compare.loc[top_langs.index]

lang_top.plot(kind='bar', figsize=(12,6), colormap="viridis")

plt.title("Programming Languages: Current vs Desired Usage")
plt.xlabel("Programming Language")
plt.ylabel("Number of Respondents")
plt.xticks(rotation=45)
```

```python
import seaborn as sns

# Select top 10 languages by combined frequency
top_langs = lang_compare.sum(axis=1).sort_values(ascending=False).head(10)
lang_top = lang_compare.loc[top_langs.index]

# Grouped bar chart
lang_top.plot(kind='bar', figsize=(12,6))
plt.title("Programming Languages: Current vs Desired Usage")
plt.xlabel("Programming Language")
plt.ylabel("Number of Respondents")
plt.xticks(rotation=45)
plt.show()
```

### Step 7: Analyze Remote Work Trends


- Visualize the distribution of RemoteWork by region using a grouped bar chart or heatmap.


```python
df_remote = df.dropna(subset=['RemoteWork', 'Country'])

# Cross-tabulation (counts)
remote_region_ct = pd.crosstab(df_remote['Country'], df_remote['RemoteWork'])

# Normalize to percentages for fair comparison
remote_region_pct = pd.crosstab(df_remote['Country'], df_remote['RemoteWork'], normalize='index') * 100

print(remote_region_ct.head())
print(remote_region_pct.head())
```

```python
import matplotlib.pyplot as plt

remote_region_pct.plot(kind='bar', figsize=(15,7), colormap="viridis")

plt.title("Remote Work Preferences by Region")
plt.xlabel("Region (Country)")
plt.ylabel("Percentage of Respondents")
plt.legend(title="Remote Work Preference", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45)
plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(12,8))
sns.heatmap(remote_region_pct, annot=True, fmt=".1f", cmap="YlGnBu")

plt.title("Remote Work Preferences by Region (Percentage of Respondents)")
plt.xlabel("Remote Work Preference")
plt.ylabel("Region (Country)")
```

### Step 8: Correlation between Job Satisfaction and Experience


- Analyze the correlation between overall job satisfaction (`JobSat`) and `YearsCodePro`.
  
- Calculate the Pearson or Spearman correlation coefficient.


```python
import pandas as pd

# Map JobSat to numeric scores
job_sat_map = {
    'Very dissatisfied': 1,
    'Dissatisfied': 2,
    'Neutral': 3,
    'Satisfied': 4,
    'Very satisfied': 5
}
df['JobSatScore'] = df['JobSat'].map(job_sat_map)

# Convert YearsCodePro to numeric (handling special cases)
def convert_years(x):
    if pd.isna(x):
        return None
    if str(x).lower().startswith("less"):
        return 0
    if str(x).lower().startswith("more"):
        return 50
    try:
        return int(x)
    except:
        return None

df['YearsCodeProNum'] = df['YearsCodePro'].apply(convert_years)

# Drop missing values
df_corr = df.dropna(subset=['JobSatScore', 'YearsCodeProNum'])
```

### Step 9: Cross-tabulation Analysis (Employment vs. Education Level)


- Analyze the relationship between employment status (`Employment`) and education level (`EdLevel`).

- **Instruction**: Create a cross-tabulation using `pd.crosstab()` and visualize it with a stacked bar plot if possible.


```python
import pandas as pd

# Drop missing values for clarity
df_ed_emp = df.dropna(subset=['Employment', 'EdLevel'])

# Cross-tabulation (counts)
ct_counts = pd.crosstab(df_ed_emp['EdLevel'], df_ed_emp['Employment'])

# Cross-tabulation (percentages within education level)
ct_percent = pd.crosstab(df_ed_emp['EdLevel'], df_ed_emp['Employment'], normalize='index') * 100

print(ct_counts.head())
print(ct_percent.head())
```

```python
import matplotlib.pyplot as plt

ct_percent.plot(kind='bar', stacked=True, figsize=(12,6), colormap="viridis")

plt.title("Employment Type Distribution by Education Level")
plt.xlabel("Education Level")
plt.ylabel("Percentage of Respondents")
plt.legend(title="Employment Type", bbox_to_anchor=(1.05, 1), loc='upper left')
plt.xticks(rotation=45)
```
